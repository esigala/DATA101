### Part 1: NumPy grade calculator
Run the cell and enter each student’s name and five scores (0–100).

```python
import numpy as np

assessments = ["Homework", "Project", "Exam 1", "Exam 2", "Final Exam"]
weights = np.array([0.15, 0.15, 0.20, 0.20, 0.30])
student_names = []
grades = np.zeros((10, 5))

for i in range(10):
    name = input(f"Enter name for student {i + 1}: ").strip()
    student_names.append(name)
    for j, assessment in enumerate(assessments):
        while True:
            try:
                score = float(input(f"  {assessment} grade for {name} (0–100): "))
                if 0 <= score <= 100:
                    grades[i, j] = score
                    break
                print("Please enter a number between 0 and 100.")
            except ValueError:
                print("Please enter a valid number.")

final_grades = grades @ weights

def letter_grade(score):
    if score >= 90: return "A"
    if score >= 80: return "B"
    if score >= 70: return "C"
    if score >= 60: return "D"
    return "F"

print("\nStudent final grades:")
for name, score in zip(student_names, final_grades):
    print(f"{name}: {score:.2f}% ({letter_grade(score)})")
```
